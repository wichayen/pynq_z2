```python
from rtlsdr import RtlSdr
import numpy as np

sdr = RtlSdr()
sdr.sample_rate = 1.92e6
sdr.gain = 'auto'

results = []

for f in np.arange(88e6, 108e6, 0.2e6):
    sdr.center_freq = f
    samples = sdr.read_samples(256*1024)
    power = np.mean(np.abs(samples)**2)
    results.append((power, f))

sdr.close()

for power, f in sorted(results, reverse=True)[:10]:
    print(f"{f/1e6:8.1f} MHz  power={power:.6f}")
```

```python
from rtlsdr import RtlSdr
import numpy as np
import matplotlib.pyplot as plt

sdr = RtlSdr()
sdr.sample_rate = 1.92e6
sdr.center_freq = 97.75e6
sdr.gain = 'auto'

samples = sdr.read_samples(256*1024)
sdr.close()

samples -= np.mean(samples)

N = len(samples)
spec = np.fft.fftshift(np.fft.fft(samples))
freq = np.fft.fftshift(np.fft.fftfreq(N, 1/sdr.sample_rate))

plt.figure(figsize=(12,4))
plt.plot(freq/1e6 + 89.2, 20*np.log10(np.abs(spec)+1e-12))
plt.xlabel("Frequency (MHz)")
plt.ylabel("Magnitude (dB)")
plt.grid()
plt.show()
```
